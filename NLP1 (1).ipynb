```python
import nltk
from nltk.tokenize import WhitespaceTokenizer, WordPunctTokenizer, TreebankWordTokenizer, TweetTokenizer, MWETokenizer
from nltk.stem import PorterStemmer, SnowballStemmer
from nltk.stem import WordNetLemmatizer

# Ensure necessary NLTK resources are downloaded
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('stopwords')

```

```python
# Sample text for tokenization
text = "Welcome to the I2IT-NLP Page. Good Morning! Who is your favourite cryptocurrency influencer? "

# Whitespace Tokenization
wt = WhitespaceTokenizer()
whitespace_tokens = wt.tokenize(text)
print("Whitespace Tokenization:", whitespace_tokens)
```

```python
# Punctuation-based Tokenization
ppt = WordPunctTokenizer()
punctuation_tokens = ppt.tokenize(text)
print("Punctuation-based Tokenization:", punctuation_tokens)

# Treebank Tokenization
treebank_tokenizer = TreebankWordTokenizer()
treebank_tokens = treebank_tokenizer.tokenize(text)
print("Treebank Tokenization:", treebank_tokens)
```

```python
# Tweet Tokenization
tweet_tokenizer = TweetTokenizer()
tweet_tokens = tweet_tokenizer.tokenize(text)
print("Tweet Tokenization:", tweet_tokens)

# Multi-Word Expression Tokenization
mwe_tokenizer = MWETokenizer([('cryptocurrency', 'influencer')])
mwe_tokens = mwe_tokenizer.tokenize(whitespace_tokens)
print("Multi-Word Expression Tokenization:", mwe_tokens)

# Stemming with Porter Stemmer
porter = PorterStemmer()
porter_stemmed = [porter.stem(word) for word in whitespace_tokens]
print("Porter Stemmer Output:", porter_stemmed)

# Stemming with Snowball Stemmer
snowball = SnowballStemmer(language='english')
snowball_stemmed = [snowball.stem(word) for word in whitespace_tokens]
print("Snowball Stemmer Output:", snowball_stemmed)

# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in whitespace_tokens]
print("Lemmatization Output:", lemmatized_words)

# Example of filtering out stop words from lemmatized words
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

# Filtering out stop words from lemmatized words
filtered_lemmatized = [word for word in lemmatized_words if word.lower() not in stop_words]
print("Filtered Lemmatized Output:", filtered_lemmatized)
```
